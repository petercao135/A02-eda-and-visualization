$$

$$
and sample covariance of two variables $X$ and $Y$,
$$
\text{cov}(X,Y) = \dfrac{1}{N} \sum_{i=1}^N (x_i - m(X))(y_i - m(Y))).
$$
Recall, the sample variance of $X$ is
$$
s^2 = \dfrac{1}{N} \sum_{i=1}^N (x_i - m(X))^2.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('./data/ForeignGifts_edu.csv')
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df["Foreign Gift Amount"].hist(bins=20, grid=False)
plt.title("Foreign gift amount")
plt.xlabel("Foreign gift amount")
plt.ylabel("Count")
plt.show()
```

```python
df["Foreign Gift Amount_adjusted"] = np.arcsinh(df["Foreign Gift Amount"])
df["Foreign Gift Amount_adjusted"].hist(bins=20, grid=False)
plt.title("Foreign gift amount after adjustment")
plt.xlabel("Foreign gift amount——arcsin_adjusted")
plt.ylabel("Count")
plt.show()
df["Foreign Gift Amount"].describe()
```

```python
df["Gift Type"].value_counts()
df["Gift Type"].value_counts(normalize=True)*100
```

```python
df_positive = df[df["Foreign Gift Amount"] > 0].copy()
df_positive["Foreign Gift Amount_kde"] = np.log(df_positive["Foreign Gift Amount"])
df_positive["Foreign Gift Amount_kde"].plot.kde()
plt.title("Foreign gift amount_kde")
plt.xlabel("Foreign gift amount")
plt.show()
```

```python
for gift_type in df_positive["Gift Type"].unique():
    df_positive[df_positive["Gift Type"] == gift_type]["Foreign Gift Amount_kde"].plot.kde(label=gift_type)
plt.title("Foreign gift amount by type")
plt.xlabel("Foreign gift amount")
plt.legend()
plt.show()
```

```python
country_count = df.groupby(["Country of Giftor"])["Foreign Gift Amount"].count()
country_count.sort_values(ascending=False).head(15)
```

```python
country_amount = df.groupby(["Country of Giftor"])["Foreign Gift Amount"].sum()
country_amount.sort_values(ascending=False).head(15)
```

```python
giftor_amount = df.groupby(["Giftor Name"])["Foreign Gift Amount"].sum()
giftor_amount.sort_values(ascending=False).head(15)
```
